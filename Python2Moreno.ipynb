Matrices

```python
#creemos una matriz de 5 filas por 5 columnas
matriz = [[1,6,11,16,21],
          [2,7,12,17,22],
          [3,8,13,18,23],
          [4,9,14,19,24],
          [5,10,15,20,25]]

#Veamos nuestra matriz
print('Impresión de una matriz')
print(matriz)

#Veamos nuestra matriz fila por fila
print("\nMatriz por filas")
print(matriz[0])#El primer corchete selecciona fila
print(matriz[1])
print(matriz[2])
print(matriz[3])
print(matriz[4])

```

Generación de matrices con ciclos o comandos automatizados

```python
#Pero que pasa si la matriz es muy grande, tiene por ejemplo 50 filas. No pondremos fila por fila como anteriormente, pero existe un formas mas eficiente, los ciclos.
for i in matriz: #le decimos que seleccione cada fila en la matriz ejecute la orden colocada despues, por un tab.
    print(i)

#Usemos un doble ciclo anidado para imprimir elemento por elemento
print("Matriz elemento por elemento")
for i in matriz:
    for j in i:
        print(j)

#Pero como se selecciona un elemento en especifico. Al igual que en las listas, se usan corchetes pero por ser de mayor dimension se utilizan dos corchetes.
#El primer corchete selecciona fila y el segundo comlumna.
print(matriz[2][2]) #seleccionemos el elemento de enmedio

#Por ultimo, para cambiar un elemento de la matriz se selecciona e iguala por el valor nuevo.
matriz[2][2] = "Nuevo elemento"
print(matriz)
```

```python
#Ejercicio 3

matriz = []
numero = 0

for i in range(10):
    fila = []

    for j in range(10):
        fila.append(numero)
        numero = numero + 1

    matriz.append(fila)

print("Matriz de 10*10: ")

for fila in matriz:
    print(fila)
```
